```python
import pandas as pd
import os
#设置显示选项，避免列太多被折叠
pd.set_option('display.max_columns', None)
#数据文件夹路径
data_path = '../data/raw'
#列出所有csv文件
files = sorted(os.listdir(data_path))
files
```

```python
for f in files:
    if f.endswith('.csv'):
        full_path = os.path.join(data_path, f)
        df = pd.read_csv(full_path)
        print('=' * 60)
        print(f'文件：{f}')
        print(f'形状:{df.shape[0]}行, {df.shape[1]}列')
        print(f'列名:{df.columns.tolist()}')
        print('缺失值数量:')
        print(df.isnull().sum())
        print()


```
